```python
import pandas as pd
import numpy as np

df = pd.read_excel("../data/raw/Telco_customer_churn.xlsx")
df.shape
```

```python
df.head
```

```python
y = df["Churn Value"]
print("Cutomers who left:", y.sum())
print("Customers who stayed", (y == 0).sum())
print(f"churn rate: {y.mean():.2%}")
```

```python
pred_nobody = np.zeros(len(y))
accuracy = (pred_nobody == y).mean()
caught = ((pred_nobody ==1) & (y ==1)).sum()
recall = caught / (y == 1).sum()
print(f"Accuracy: {accuracy:.2%}")
print(f"Churners caught: {caught} out of {y.sum()} ( recall = {recall:.0%})")
```

```python
print(df.groupby("Churn Label")["Churn Score"].agg(["min","max","mean"]))
print()
print(pd.crosstab(df["Churn Label"], df["Churn Reason"].notna()))
```

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors="coerce")
print("Blank values found:", df["Total Charges"].isna().sum())
assert (df.loc[df["Total Charges"].isna(), "Tenure Months"] == 0).all()
df["Total Charges"] = df["Total Charges"].fillna(0)
print("Blank values left:", df["Total Charges"].isna().sum())
```

```python
df = df.drop(columns=["Count", "Country", "State", "Lat Long"])
df.to_csv("../data/processed/telco_clean.csv", index = False)
print(df.shape)
```
